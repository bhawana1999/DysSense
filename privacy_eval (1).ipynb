```python
import re
import numpy as np
import torch
import soundfile as sf
import torchaudio

from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC
from jiwer import wer

# -----------------------------
# Normalization
# -----------------------------
def normalize_words(text: str) -> str:
    text = text.lower()
    text = re.sub(r"[^a-z0-9' ]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def normalize_chars(text: str) -> str:
    # for CER: keep only alnum + space, collapse spaces
    text = text.lower()
    text = re.sub(r"[^a-z0-9 ]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# -----------------------------
# CER (character error rate)
# -----------------------------
def cer(ref: str, hyp: str) -> float:
    ref = normalize_chars(ref)
    hyp = normalize_chars(hyp)
    if len(ref) == 0:
        return np.nan

    # classic Levenshtein distance on characters
    n, m = len(ref), len(hyp)
    dp = [[0]*(m+1) for _ in range(n+1)]
    for i in range(n+1): dp[i][0] = i
    for j in range(m+1): dp[0][j] = j

    for i in range(1, n+1):
        for j in range(1, m+1):
            if ref[i-1] == hyp[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(dp[i-1][j-1], dp[i-1][j], dp[i][j-1])

    return dp[n][m] / n

# -----------------------------
# Audio loading (NO torchaudio.load)
# -----------------------------
def load_audio_16k(path: str):
    wav, sr = sf.read(path)  # wav is numpy
    if wav.ndim > 1:
        wav = wav.mean(axis=1)  # mono

    wav = torch.tensor(wav, dtype=torch.float32)

    # resample if needed
    if sr != 16000:
        wav = torchaudio.functional.resample(wav, sr, 16000)
        sr = 16000

    return wav, sr

# -----------------------------
# wav2vec2 ASR
# -----------------------------
def build_asr(model_name="facebook/wav2vec2-base-960h"):
    device = "cuda" if torch.cuda.is_available() else "cpu"
    processor = Wav2Vec2Processor.from_pretrained(model_name)
    model = Wav2Vec2ForCTC.from_pretrained(model_name).to(device)
    model.eval()
    return processor, model, device

@torch.inference_mode()
def transcribe(audio_path: str, processor, model, device: str) -> str:
    wav, sr = load_audio_16k(audio_path)
    inputs = processor(wav.numpy(), sampling_rate=sr, return_tensors="pt", padding=True)
    input_values = inputs.input_values.to(device)

    logits = model(input_values).logits
    pred_ids = torch.argmax(logits, dim=-1)
    text = processor.batch_decode(pred_ids)[0]
    return text

# -----------------------------
# Evaluation
# -----------------------------
def evaluate_set(audio_files, transcripts, processor, model, device):
    wers, cers, hyps = [], [], []

    for audio, ref in zip(audio_files, transcripts):
        hyp = transcribe(audio, processor, model, device)
        hyps.append(hyp)

        ref_w = normalize_words(ref)
        hyp_w = normalize_words(hyp)

        wers.append(wer(ref_w, hyp_w) if ref_w else np.nan)
        cers.append(cer(ref, hyp))

    return np.array(wers), np.array(cers), hyps

# -----------------------------
# RUN (expects your 4 lists exist)
# -----------------------------
# audio_files_1, transcripts_1, audio_files_2, transcripts_2

print (audio_files_1, transcripts_1, audio_files_2, transcripts_2)

processor, model, device = build_asr("facebook/wav2vec2-base-960h")

wers_1, cers_1, hyps_1 = evaluate_set(audio_files_1, transcripts_1, processor, model, device)
wers_2, cers_2, hyps_2 = evaluate_set(audio_files_2, transcripts_2, processor, model, device)

print("\n=== wav2vec2 Privacy Leakage ===")
print(f"Device: {device}")
print(f"Set 1 (speech):  WER mean={np.nanmean(wers_1):.3f} | CER mean={np.nanmean(cers_1):.3f}")
print(f"Set 2 (private): WER mean={np.nanmean(wers_2):.3f} | CER mean={np.nanmean(cers_2):.3f}")
print(f"ΔWER (private-speech): {np.nanmean(wers_2 - wers_1):.3f}")
print(f"ΔCER (private-speech): {np.nanmean(cers_2 - cers_1):.3f}")
```

```python
import re, numpy as np, torch, soundfile as sf, torchaudio
from transformers import Wav2Vec2Processor, Wav2Vec2ForCTC
from jiwer import wer

def normalize_words(t):
    t = t.lower()
    t = re.sub(r"[^a-z0-9' ]+", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

def load_audio_16k(path):
    wav, sr = sf.read(path)
    if wav.ndim > 1:
        wav = wav.mean(axis=1)
    wav = torch.tensor(wav, dtype=torch.float32)
    if sr != 16000:
        wav = torchaudio.functional.resample(wav, sr, 16000)
    return wav

@torch.inference_mode()
def transcribe_batch(paths, processor, model, device):
    wavs = [load_audio_16k(p) for p in paths]
    inputs = processor(
        [w.numpy() for w in wavs],
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    )
    input_values = inputs.input_values.to(device)
    logits = model(input_values).logits
    pred_ids = torch.argmax(logits, dim=-1)
    return processor.batch_decode(pred_ids)

def evaluate_set_batched(audio_files, transcripts, processor, model, device, batch_size=16):
    wers = []
    hyps = []

    for i in range(0, len(audio_files), batch_size):
        batch_paths = audio_files[i:i+batch_size]
        batch_refs  = transcripts[i:i+batch_size]

        batch_hyps = transcribe_batch(batch_paths, processor, model, device)
        hyps.extend(batch_hyps)

        for ref, hyp in zip(batch_refs, batch_hyps):
            ref_n = normalize_words(ref)
            hyp_n = normalize_words(hyp)
            wers.append(wer(ref_n, hyp_n) if ref_n else np.nan)

    return np.array(wers), hyps

# ---- build model
device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h").to(device)
model.eval()

# ---- run both sets (WER only for speed; add CER later if needed)
wers_1, hyps_1 = evaluate_set_batched(audio_files_1, transcripts_1, processor, model, device, batch_size=16)
wers_2, hyps_2 = evaluate_set_batched(audio_files_2, transcripts_2, processor, model, device, batch_size=16)

print("Set1 WER mean:", np.nanmean(wers_1))
print("Set2 WER mean:", np.nanmean(wers_2))
print("ΔWER mean:", np.nanmean(wers_2 - wers_1))
```

```python
# ============================================================
# ASR-BASED SPEECH PRIVACY EVALUATION
# Whisper + Wav2Vec2 + ASR Agreement
# ============================================================

!pip install -q openai-whisper transformers jiwer g2p_en

import os
import re
import glob
import torch
import whisper
import numpy as np
import pandas as pd
import librosa
import soundfile as sf

from g2p_en import G2p
from transformers import Wav2Vec2ForCTC, Wav2Vec2Processor
from jiwer import wer


# ============================================================
# Severity labels
# ============================================================

DYSARTHRIC_SPEAKER_INFO = {
    "M01": "very_low", "M04": "very_low",
    "M05": "medium", "M06": "low", "M07": "low",
    "M08": "high", "M09": "high", "M10": "high",
    "F02": "low", "F03": "very_low", "F04": "medium",
    "F05": "high", "M11": "medium", "M12": "very_low",
    "M14": "high", "M16": "low",
}


def get_severity(speaker_id):
    if speaker_id.startswith("C"):
        return "control"
    return DYSARTHRIC_SPEAKER_INFO.get(speaker_id, "unknown")


# ============================================================
# Transcript parsing
# ============================================================

def extract_transcript_from_lab(file_path, utterance_id):
    with open(file_path, "r") as f:
        content = f.read()

    index = content.find(utterance_id)

    if index == -1:
        return None

    start_index = index + len(utterance_id) + 5
    dot_index = content.find(".", start_index)

    if dot_index == -1:
        return None

    text = content[start_index:dot_index].strip()

    text = text.replace("..", "")
    text = text.replace(".", "")
    text = text.strip()

    return text


def build_uaspeech_asr_metadata(
    control_dir,
    fm_dir,
    control_transcript_dir,
    fm_transcript_dir,
    selected_mic=None,
    max_files_per_speaker=None,
):
    rows = []

    # controls
    for speaker_id in sorted(os.listdir(control_dir)):
        speaker_audio_dir = os.path.join(control_dir, speaker_id)
        print (speaker_audio_dir, os.path.isdir(speaker_audio_dir), speaker_id.startswith("C"))

        if not os.path.isdir(speaker_audio_dir):
            continue

        if not speaker_id.startswith("C"):
            continue

        transcript_file = os.path.join(
            control_transcript_dir,
            speaker_id,
            f"{speaker_id}_word.mlf"
        )

        print ("t", transcript_file, os.path.exists(transcript_file))

        if not os.path.exists(transcript_file):
            continue

        wavs = sorted(glob.glob(os.path.join(speaker_audio_dir, "*.wav")))

        print ( len(wavs))

        if selected_mic is not None:
            wavs = [w for w in wavs if f"_{selected_mic}.wav" in w]

        if max_files_per_speaker is not None:
            wavs = wavs[:max_files_per_speaker]

        print ("wavs", len(wavs))

        for wav_path in wavs:
            utt_id = os.path.splitext(os.path.basename(wav_path))[0]
            text = extract_transcript_from_lab(transcript_file, utt_id)
            print (utt_id, text)

            if text is None:
                continue

            rows.append({
                "audio_path": wav_path,
                "filename": os.path.basename(wav_path),
                "utterance_id": utt_id,
                "speaker_id": speaker_id,
                "severity": "control",
                "reference": text,
            })

    # dysarthric
    for speaker_id in sorted(os.listdir(fm_dir)):
        speaker_audio_dir = os.path.join(fm_dir, speaker_id)

        if not os.path.isdir(speaker_audio_dir):
            continue

        if speaker_id not in DYSARTHRIC_SPEAKER_INFO:
            continue

        transcript_file = os.path.join(
            fm_transcript_dir,
            speaker_id,
            f"{speaker_id}_word.mlf"
        )

        print ("t", transcript_file, os.path.exists(transcript_file), speaker_audio_dir)

        if not os.path.exists(transcript_file):
            continue




        wavs = sorted(glob.glob(os.path.join(speaker_audio_dir, "*.wav")))
        print ("wavs", len(wavs))

        if selected_mic is not None:
            wavs = [w for w in wavs if f"_{selected_mic}.wav" in w]

        if max_files_per_speaker is not None:
            wavs = wavs[:max_files_per_speaker]

        for wav_path in wavs:
            utt_id = os.path.splitext(os.path.basename(wav_path))[0]
            text = extract_transcript_from_lab(transcript_file, utt_id)

            print (utt_id, text)


            if text is None:
                continue

            rows.append({
                "audio_path": wav_path,
                "filename": os.path.basename(wav_path),
                "utterance_id": utt_id,
                "speaker_id": speaker_id,
                "severity": get_severity(speaker_id),
                "reference": text,
            })

    df = pd.DataFrame(rows)

    print("ASR metadata shape:", df.shape)
    print(df["severity"].value_counts())
    print("num speakers:", df["speaker_id"].nunique())

    return df


# ============================================================
# Text normalization + metrics
# ============================================================

def normalize_text(text):
    text = str(text).upper()
    text = re.sub(r"[^A-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def levenshtein(a, b):
    n, m = len(a), len(b)
    dp = np.zeros((n + 1, m + 1), dtype=int)

    for i in range(n + 1):
        dp[i, 0] = i

    for j in range(m + 1):
        dp[0, j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if a[i - 1] == b[j - 1] else 1
            dp[i, j] = min(
                dp[i - 1, j] + 1,
                dp[i, j - 1] + 1,
                dp[i - 1, j - 1] + cost,
            )

    return dp[n, m]


def cer_score(ref, hyp):
    ref = normalize_text(ref).replace(" ", "")
    hyp = normalize_text(hyp).replace(" ", "")

    if len(ref) == 0:
        return 0.0 if len(hyp) == 0 else 1.0

    return levenshtein(list(ref), list(hyp)) / len(ref)


g2p = G2p()

def text_to_phonemes(text):
    text = normalize_text(text)
    phones = g2p(text)

    phones = [
        p for p in phones
        if p.strip() != "" and p != " "
    ]

    return phones


def per_score(ref, hyp):
    """
    Approximate PER using G2P phonemization of reference and ASR hypothesis.
    This is not a forced-alignment PER, but useful as a privacy leakage proxy.
    """
    ref_p = text_to_phonemes(ref)
    hyp_p = text_to_phonemes(hyp)

    if len(ref_p) == 0:
        return 0.0 if len(hyp_p) == 0 else 1.0

    return levenshtein(ref_p, hyp_p) / len(ref_p)


def wer_score(ref, hyp):
    ref = normalize_text(ref)
    hyp = normalize_text(hyp)

    if len(ref.split()) == 0:
        return 0.0 if len(hyp.split()) == 0 else 1.0

    return wer(ref, hyp)


def asr_disagreement_score(hyp1, hyp2):
    """
    Agreement proxy between Whisper and Wav2Vec2.
    Higher = more disagreement.
    Uses CER between their outputs.
    """
    return cer_score(hyp1, hyp2)


# ============================================================
# Audio + Anti-VAD
# ============================================================

def load_audio_16k(path):
    y, sr = librosa.load(path, sr=16000, mono=True)
    y = y.astype(np.float32)
    y = y / (np.max(np.abs(y)) + 1e-8)
    return y, 16000


print("Loading Silero VAD...")
silero_model, silero_utils = torch.hub.load(
    repo_or_dir="snakers4/silero-vad",
    model="silero_vad",
    force_reload=False
)
get_speech_timestamps = silero_utils[0]


def silero_antivad(y, sr=16000, threshold=0.8):
    timestamps = get_speech_timestamps(
        torch.from_numpy(y),
        silero_model,
        sampling_rate=sr,
        threshold=threshold
    )

    mask = np.zeros(len(y), dtype=bool)

    for seg in timestamps:
        mask[int(seg["start"]):int(seg["end"])] = True

    y_private = y.copy()
    y_private[mask] = 0.0

    return y_private.astype(np.float32)


def create_audio_representation(y, sr, representation):
    if representation == "raw":
        return y

    threshold = float(representation.replace("antivad_", ""))
    return silero_antivad(y, sr, threshold=threshold)


# ============================================================
# Load ASR models
# ============================================================

def load_asr_models(
    whisper_model_name="base",
    wav2vec_model_name="facebook/wav2vec2-base-960h"
):
    print("Loading Whisper:", whisper_model_name)
    whisper_model = whisper.load_model(whisper_model_name)

    print("Loading Wav2Vec2:", wav2vec_model_name)
    wav2vec_processor = Wav2Vec2Processor.from_pretrained(wav2vec_model_name)
    wav2vec_model = Wav2Vec2ForCTC.from_pretrained(wav2vec_model_name)

    device = "cuda" if torch.cuda.is_available() else "cpu"
    wav2vec_model = wav2vec_model.to(device)

    return whisper_model, wav2vec_processor, wav2vec_model, device


def transcribe_whisper(y, sr, whisper_model, tmp_path):
    sf.write(tmp_path, y, sr)
    result = whisper_model.transcribe(tmp_path, fp16=torch.cuda.is_available())
    return normalize_text(result.get("text", ""))


def transcribe_wav2vec(y, sr, processor, model, device):
    inputs = processor(
        y,
        sampling_rate=sr,
        return_tensors="pt",
        padding=True
    )

    input_values = inputs.input_values.to(device)

    with torch.no_grad():
        logits = model(input_values).logits

    pred_ids = torch.argmax(logits, dim=-1)
    transcript = processor.batch_decode(pred_ids)[0]

    return normalize_text(transcript)


# ============================================================
# Main ASR privacy evaluation
# ============================================================

def evaluate_asr_privacy(
    metadata,
    output_dir,
    thresholds=(0.4, 0.5, 0.6, 0.7, 0.8, 0.9),
    whisper_model_name="base",
    wav2vec_model_name="facebook/wav2vec2-base-960h",
    max_files_per_severity=None,
):
    os.makedirs(output_dir, exist_ok=True)
    tmp_dir = os.path.join(output_dir, "asr_tmp")
    os.makedirs(tmp_dir, exist_ok=True)

    whisper_model, processor, wav2vec_model, device = load_asr_models(
        whisper_model_name=whisper_model_name,
        wav2vec_model_name=wav2vec_model_name
    )

    reps = ["raw"] + [f"antivad_{t}" for t in thresholds]

    if max_files_per_severity is not None:
        eval_df = (
            metadata
            .groupby("severity", group_keys=False)
            .head(max_files_per_severity)
            .copy()
        )
    else:
        eval_df = metadata.copy()

    print("Evaluation size:", eval_df.shape)
    print(eval_df["severity"].value_counts())

    rows = []

    for idx, row in eval_df.iterrows():
        print(f"\n[{len(rows)}] {row['filename']} | {row['severity']} | ref={row['reference']}")

        y, sr = load_audio_16k(row["audio_path"])

        for rep in reps:
            print("  Representation:", rep)

            y_rep = create_audio_representation(y, sr, rep)

            tmp_path = os.path.join(
                tmp_dir,
                f"{idx}_{rep}.wav"
            )

            whisper_hyp = transcribe_whisper(
                y_rep, sr, whisper_model, tmp_path
            )

            wav2vec_hyp = transcribe_wav2vec(
                y_rep, sr, processor, wav2vec_model, device
            )

            ref = row["reference"]

            whisper_cer = cer_score(ref, whisper_hyp)
            whisper_wer = wer_score(ref, whisper_hyp)
            whisper_per = per_score(ref, whisper_hyp)

            wav2vec_cer = cer_score(ref, wav2vec_hyp)
            wav2vec_wer = wer_score(ref, wav2vec_hyp)
            wav2vec_per = per_score(ref, wav2vec_hyp)

            disagreement = asr_disagreement_score(
                whisper_hyp,
                wav2vec_hyp
            )

            rows.append({
                "audio_path": row["audio_path"],
                "filename": row["filename"],
                "speaker_id": row["speaker_id"],
                "severity": row["severity"],
                "reference": ref,
                "representation": rep,

                "whisper_hyp": whisper_hyp,
                "wav2vec_hyp": wav2vec_hyp,

                "whisper_cer": whisper_cer,
                "whisper_wer": whisper_wer,
                "whisper_per": whisper_per,

                "wav2vec_cer": wav2vec_cer,
                "wav2vec_wer": wav2vec_wer,
                "wav2vec_per": wav2vec_per,

                "asr_disagreement_cer": disagreement,

                "whisper_decoded_words": len(whisper_hyp.split()),
                "wav2vec_decoded_words": len(wav2vec_hyp.split()),

                "residual_energy_ratio": float(
                    np.sum(y_rep ** 2) / (np.sum(y ** 2) + 1e-8)
                ),
                "zero_fraction": float(np.mean(np.abs(y_rep) < 1e-8)),
            })

            print("    Whisper:", whisper_hyp, "| CER:", whisper_cer)
            print("    W2V    :", wav2vec_hyp, "| CER:", wav2vec_cer)
            print("    Disagreement:", disagreement)

    results_df = pd.DataFrame(rows)

    out_path = os.path.join(output_dir, "asr_privacy_full_results.csv")
    results_df.to_csv(out_path, index=False)

    print("Saved full results:", out_path)

    summary = (
        results_df
        .groupby(["representation"])
        [[
            "whisper_cer", "whisper_wer", "whisper_per",
            "wav2vec_cer", "wav2vec_wer", "wav2vec_per",
            "asr_disagreement_cer",
            "whisper_decoded_words", "wav2vec_decoded_words",
            "residual_energy_ratio", "zero_fraction"
        ]]
        .mean()
        .reset_index()
    )

    summary_path = os.path.join(output_dir, "asr_privacy_summary_by_threshold.csv")
    summary.to_csv(summary_path, index=False)

    severity_summary = (
        results_df
        .groupby(["severity", "representation"])
        [[
            "whisper_cer", "whisper_wer", "whisper_per",
            "wav2vec_cer", "wav2vec_wer", "wav2vec_per",
            "asr_disagreement_cer",
            "whisper_decoded_words", "wav2vec_decoded_words",
            "residual_energy_ratio", "zero_fraction"
        ]]
        .mean()
        .reset_index()
    )

    severity_path = os.path.join(output_dir, "asr_privacy_summary_by_severity.csv")
    severity_summary.to_csv(severity_path, index=False)

    print("Saved summary:", summary_path)
    print("Saved severity summary:", severity_path)

    return results_df, summary, severity_summary
```

```python
# ============================================================
# ASR PRIVACY ANALYSIS + PAPER PLOTS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr

OUT_DIR = "outputs_asr_privacy/figures"
os.makedirs(OUT_DIR, exist_ok=True)

asr_results = pd.read_csv("outputs_asr_privacy/asr_privacy_full_results.csv")
asr_summary = pd.read_csv("outputs_asr_privacy/asr_privacy_summary_by_threshold.csv")
asr_severity = pd.read_csv("outputs_asr_privacy/asr_privacy_summary_by_severity.csv")

# ------------------------------------------------------------
# Paper plot style
# ------------------------------------------------------------

plt.rcParams.update({
    "font.size": 17,
    "axes.titlesize": 20,
    "axes.labelsize": 18,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "legend.fontsize": 14,
    "axes.linewidth": 1.8,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


def clean_representation(rep):
    if rep == "raw":
        return "Raw"
    return rep.replace("antivad_", "")


def savefig(path):
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Saved:", path)


# ============================================================
# 1. FIX DISAGREEMENT ISSUE
# ============================================================

def safe_text(x):
    if pd.isna(x):
        return ""
    return str(x).strip()


def decoded_words(x):
    x = safe_text(x)
    if len(x) == 0:
        return 0
    return len(x.split())


def normalized_edit_distance_maxlen(a, b):
    """
    Better disagreement metric.
    Uses max(len(a), len(b)) denominator.
    If both ASRs output nothing, disagreement = 0.
    If one is empty and the other is non-empty, disagreement = 1.
    """
    a = safe_text(a).replace(" ", "")
    b = safe_text(b).replace(" ", "")

    if len(a) == 0 and len(b) == 0:
        return 0.0

    denom = max(len(a), len(b), 1)

    n, m = len(a), len(b)
    dp = np.zeros((n + 1, m + 1), dtype=int)

    for i in range(n + 1):
        dp[i, 0] = i
    for j in range(m + 1):
        dp[0, j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if a[i - 1] == b[j - 1] else 1
            dp[i, j] = min(
                dp[i - 1, j] + 1,
                dp[i, j - 1] + 1,
                dp[i - 1, j - 1] + cost,
            )

    return dp[n, m] / denom


asr_results["whisper_words"] = asr_results["whisper_hyp"].apply(decoded_words)
asr_results["wav2vec_words"] = asr_results["wav2vec_hyp"].apply(decoded_words)

asr_results["joint_decoded_words"] = (
    asr_results["whisper_words"] + asr_results["wav2vec_words"]
) / 2

asr_results["both_empty"] = (
    (asr_results["whisper_words"] == 0) &
    (asr_results["wav2vec_words"] == 0)
)

asr_results["asr_disagreement_fixed"] = asr_results.apply(
    lambda r: normalized_edit_distance_maxlen(
        r["whisper_hyp"],
        r["wav2vec_hyp"]
    ),
    axis=1
)

# Clip metrics for easier plotting.
# Keep raw columns untouched; create clipped versions.
for col in [
    "whisper_cer", "whisper_per", "whisper_wer",
    "wav2vec_cer", "wav2vec_per", "wav2vec_wer",
]:
    asr_results[col + "_clip"] = asr_results[col].clip(0, 1)

asr_results.to_csv(
    "outputs_asr_privacy/asr_privacy_full_results_fixed.csv",
    index=False
)

print("Both ASRs empty count:")
print(asr_results.groupby("representation")["both_empty"].mean())


# ============================================================
# 2. RECOMPUTE SUMMARY WITH FIXED METRICS
# ============================================================

summary_fixed = (
    asr_results
    .groupby("representation")
    [[
        "whisper_cer_clip", "whisper_per_clip", "whisper_wer_clip",
        "wav2vec_cer_clip", "wav2vec_per_clip", "wav2vec_wer_clip",
        "asr_disagreement_fixed",
        "joint_decoded_words",
        "whisper_words", "wav2vec_words",
        "both_empty",
        "residual_energy_ratio", "zero_fraction"
    ]]
    .mean()
    .reset_index()
)

severity_fixed = (
    asr_results
    .groupby(["severity", "representation"])
    [[
        "whisper_cer_clip", "whisper_per_clip", "whisper_wer_clip",
        "wav2vec_cer_clip", "wav2vec_per_clip", "wav2vec_wer_clip",
        "asr_disagreement_fixed",
        "joint_decoded_words",
        "whisper_words", "wav2vec_words",
        "both_empty",
        "residual_energy_ratio", "zero_fraction"
    ]]
    .mean()
    .reset_index()
)

summary_fixed.to_csv(
    "outputs_asr_privacy/asr_privacy_summary_fixed_by_threshold.csv",
    index=False
)

severity_fixed.to_csv(
    "outputs_asr_privacy/asr_privacy_summary_fixed_by_severity.csv",
    index=False
)

display(summary_fixed)
display(severity_fixed)


# ============================================================
# 3. CORRELATION: DISAGREEMENT VS CER/PER
# ============================================================

def corr_row(df, x_col, y_col, label):
    temp = df[[x_col, y_col]].replace([np.inf, -np.inf], np.nan).dropna()

    if len(temp) < 3:
        return None

    pearson_r, pearson_p = pearsonr(temp[x_col], temp[y_col])
    spearman_r, spearman_p = spearmanr(temp[x_col], temp[y_col])

    return {
        "comparison": label,
        "n": len(temp),
        "pearson_r": pearson_r,
        "pearson_p": pearson_p,
        "spearman_r": spearman_r,
        "spearman_p": spearman_p,
    }


corrs = []
for y_col, label in [
    ("whisper_cer_clip", "Disagreement vs Whisper CER"),
    ("whisper_per_clip", "Disagreement vs Whisper PER"),
    ("wav2vec_cer_clip", "Disagreement vs Wav2Vec CER"),
    ("wav2vec_per_clip", "Disagreement vs Wav2Vec PER"),
]:
    corrs.append(corr_row(asr_results, "asr_disagreement_fixed", y_col, label))

corr_df = pd.DataFrame([c for c in corrs if c is not None])
corr_df.to_csv(
    "outputs_asr_privacy/asr_disagreement_correlations_fixed.csv",
    index=False
)

display(corr_df)


# Per-severity correlations
severity_corrs = []

for sev in sorted(asr_results["severity"].unique()):
    sev_df = asr_results[asr_results["severity"] == sev]

    for y_col, label in [
        ("whisper_cer_clip", "Disagreement vs Whisper CER"),
        ("whisper_per_clip", "Disagreement vs Whisper PER"),
        ("wav2vec_cer_clip", "Disagreement vs Wav2Vec CER"),
        ("wav2vec_per_clip", "Disagreement vs Wav2Vec PER"),
    ]:
        c = corr_row(sev_df, "asr_disagreement_fixed", y_col, label)
        if c is not None:
            c["severity"] = sev
            severity_corrs.append(c)

severity_corr_df = pd.DataFrame(severity_corrs)
severity_corr_df.to_csv(
    "outputs_asr_privacy/asr_disagreement_correlations_fixed_by_severity.csv",
    index=False
)

display(severity_corr_df)


# ============================================================
# 4. PLOT: CER/PER BY THRESHOLD
# ============================================================

def plot_asr_error_by_threshold(summary_df):
    df = summary_df.copy()
    df["rep_label"] = df["representation"].apply(clean_representation)

    order = ["Raw", "0.4", "0.5", "0.6", "0.7", "0.8", "0.9"]
    df["rep_label"] = pd.Categorical(df["rep_label"], categories=order, ordered=True)
    df = df.sort_values("rep_label")

    plt.figure(figsize=(9, 5))

    plt.plot(df["rep_label"], df["whisper_cer_clip"], marker="o", linewidth=2.5, label="Whisper CER")
    plt.plot(df["rep_label"], df["wav2vec_cer_clip"], marker="o", linewidth=2.5, label="Wav2Vec CER")
    plt.plot(df["rep_label"], df["whisper_per_clip"], marker="s", linewidth=2.5, label="Whisper PER")
    plt.plot(df["rep_label"], df["wav2vec_per_clip"], marker="s", linewidth=2.5, label="Wav2Vec PER")

    for _, r in df.iterrows():
        plt.text(r["rep_label"], r["whisper_cer_clip"] + 0.02, f"{r['whisper_cer_clip']:.2f}",
                 ha="center", fontsize=11, fontweight="bold")

    plt.ylabel("Error Rate")
    plt.xlabel("Representation / Anti-VAD Threshold")
    plt.ylim(0, 1.08)
    plt.title("ASR Error Increases Under Anti-VAD Filtering")
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.legend(frameon=False, ncol=2)

    savefig(os.path.join(OUT_DIR, "fig_asr_cer_per_by_threshold.pdf"))


plot_asr_error_by_threshold(summary_fixed)


# ============================================================
# 5. PLOT: DECODED WORDS BY THRESHOLD
# ============================================================

def plot_decoded_words(summary_df):
    df = summary_df.copy()
    df["rep_label"] = df["representation"].apply(clean_representation)

    order = ["Raw", "0.4", "0.5", "0.6", "0.7", "0.8", "0.9"]
    df["rep_label"] = pd.Categorical(df["rep_label"], categories=order, ordered=True)
    df = df.sort_values("rep_label")

    plt.figure(figsize=(8.5, 5))

    plt.plot(df["rep_label"], df["joint_decoded_words"], marker="o", linewidth=3, label="Avg. decoded words")

    for _, r in df.iterrows():
        plt.text(
            r["rep_label"],
            r["joint_decoded_words"] + 0.05,
            f"{r['joint_decoded_words']:.2f}",
            ha="center",
            fontsize=12,
            fontweight="bold"
        )

    plt.ylabel("Mean Decoded Words")
    plt.xlabel("Representation / Anti-VAD Threshold")
    plt.title("Decoded Linguistic Content Decreases After Anti-VAD")
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.ylim(bottom=0)
    plt.legend(frameon=False)

    savefig(os.path.join(OUT_DIR, "fig_asr_decoded_words_by_threshold.pdf"))


plot_decoded_words(summary_fixed)


# ============================================================
# 6. PLOT: ASR DISAGREEMENT BY THRESHOLD
# ============================================================

def plot_asr_disagreement(summary_df):
    df = summary_df.copy()
    df["rep_label"] = df["representation"].apply(clean_representation)

    order = ["Raw", "0.4", "0.5", "0.6", "0.7", "0.8", "0.9"]
    df["rep_label"] = pd.Categorical(df["rep_label"], categories=order, ordered=True)
    df = df.sort_values("rep_label")

    plt.figure(figsize=(8.5, 5))

    plt.plot(df["rep_label"], df["asr_disagreement_fixed"], marker="o", linewidth=3)

    for _, r in df.iterrows():
        plt.text(
            r["rep_label"],
            r["asr_disagreement_fixed"] + 0.02,
            f"{r['asr_disagreement_fixed']:.2f}",
            ha="center",
            fontsize=12,
            fontweight="bold"
        )

    plt.ylabel("ASR Disagreement")
    plt.xlabel("Representation / Anti-VAD Threshold")
    plt.ylim(0, 1.05)
    plt.title("Whisper--Wav2Vec Disagreement Across Thresholds")
    plt.grid(axis="y", linestyle="--", alpha=0.4)

    savefig(os.path.join(OUT_DIR, "fig_asr_disagreement_by_threshold.pdf"))


plot_asr_disagreement(summary_fixed)


# ============================================================
# 7. PLOT: SEVERITY-WISE PRIVACY LEAKAGE
# ============================================================

def plot_severity_asr_leakage(severity_df, metric="joint_decoded_words"):
    df = severity_df.copy()
    df["rep_label"] = df["representation"].apply(clean_representation)

    severity_order = ["control", "high", "medium", "low", "very_low"]
    rep_order = ["Raw", "0.7", "0.8", "0.9"]

    df = df[df["rep_label"].isin(rep_order)]
    df["severity"] = pd.Categorical(df["severity"], categories=severity_order, ordered=True)
    df = df.sort_values("severity")

    pivot = df.pivot(index="severity", columns="rep_label", values=metric)
    pivot = pivot[rep_order]

    ax = pivot.plot(
        kind="bar",
        figsize=(10, 5),
        width=0.72,
        edgecolor="black",
        linewidth=1.4
    )

    for container in ax.containers:
        ax.bar_label(container, fmt="%.2f", fontsize=9, fontweight="bold", padding=2)

    plt.ylabel(metric.replace("_", " ").title())
    plt.xlabel("Severity")
    plt.title(f"Severity-wise ASR Privacy Leakage: {metric.replace('_', ' ').title()}")
    plt.xticks(rotation=20, ha="right")
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.legend(title="Representation", frameon=False, ncol=4)

    savefig(os.path.join(OUT_DIR, f"fig_severity_{metric}.pdf"))


plot_severity_asr_leakage(severity_fixed, metric="joint_decoded_words")
plot_severity_asr_leakage(severity_fixed, metric="whisper_cer_clip")
plot_severity_asr_leakage(severity_fixed, metric="wav2vec_cer_clip")


# ============================================================
# 8. PLOT: DISAGREEMENT VS CER/PER SCATTER
# ============================================================

def plot_disagreement_scatter(asr_df, y_col="whisper_cer_clip"):
    df = asr_df.copy()

    plt.figure(figsize=(7, 5))
    plt.scatter(
        df["asr_disagreement_fixed"],
        df[y_col],
        s=45,
        edgecolor="black",
        linewidth=0.8,
        alpha=0.75
    )

    r, p = spearmanr(df["asr_disagreement_fixed"], df[y_col])

    plt.xlabel("Whisper--Wav2Vec Disagreement")
    plt.ylabel(y_col.replace("_", " ").title())
    plt.title(f"ASR Disagreement as Privacy Proxy\nSpearman r={r:.2f}, p={p:.3g}")
    plt.grid(True, linestyle="--", alpha=0.35)
    plt.xlim(-0.03, 1.03)
    plt.ylim(-0.03, 1.03)

    savefig(os.path.join(OUT_DIR, f"fig_disagreement_vs_{y_col}.pdf"))


plot_disagreement_scatter(asr_results, "whisper_cer_clip")
plot_disagreement_scatter(asr_results, "whisper_per_clip")
plot_disagreement_scatter(asr_results, "wav2vec_cer_clip")
plot_disagreement_scatter(asr_results, "wav2vec_per_clip")


# ============================================================
# 9. UTILITY-PRIVACY TRADEOFF
# ============================================================

# Fill these with your utility results.
# Example values from your current 3-class results:
utility_3class = {
    "raw": 0.734233,              # macro-F1 or balanced accuracy
    "antivad_0.7": 0.714448,
    "antivad_0.8": 0.724567,
    "antivad_0.9": 0.734233,
}

def plot_utility_privacy_tradeoff(summary_df, utility_dict):
    df = summary_df.copy()
    df = df[df["representation"].isin(utility_dict.keys())].copy()

    df["utility"] = df["representation"].map(utility_dict)
    df["privacy_leakage"] = df["joint_decoded_words"]
    df["rep_label"] = df["representation"].apply(clean_representation)

    plt.figure(figsize=(7, 5))

    plt.scatter(
        df["privacy_leakage"],
        df["utility"],
        s=130,
        edgecolor="black",
        linewidth=1.4
    )

    for _, r in df.iterrows():
        plt.text(
            r["privacy_leakage"] + 0.02,
            r["utility"],
            r["rep_label"],
            fontsize=13,
            fontweight="bold"
        )

    plt.xlabel("Speech Leakage Proxy: Mean Decoded Words")
    plt.ylabel("Dysarthria Utility")
    plt.title("Privacy--Utility Tradeoff")
    plt.grid(True, linestyle="--", alpha=0.35)

    savefig(os.path.join(OUT_DIR, "fig_privacy_utility_tradeoff.pdf"))


plot_utility_privacy_tradeoff(summary_fixed, utility_3class)
```

```python
# ============================================================
# REAL-TIME PRIVACY METRIC VS CER/PER
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr

OUT_DIR = "outputs_asr_privacy/figures"
os.makedirs(OUT_DIR, exist_ok=True)

asr_results = pd.read_csv("outputs_asr_privacy/asr_privacy_full_results.csv")


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def safe_text(x):
    if pd.isna(x):
        return ""
    return str(x).strip()


def count_words(x):
    x = safe_text(x)
    if len(x) == 0:
        return 0
    return len(x.split())


def normalized_edit_distance_maxlen(a, b):
    a = safe_text(a).replace(" ", "")
    b = safe_text(b).replace(" ", "")

    if len(a) == 0 and len(b) == 0:
        return 0.0

    n, m = len(a), len(b)
    dp = np.zeros((n + 1, m + 1), dtype=int)

    for i in range(n + 1):
        dp[i, 0] = i

    for j in range(m + 1):
        dp[0, j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if a[i - 1] == b[j - 1] else 1
            dp[i, j] = min(
                dp[i - 1, j] + 1,
                dp[i, j - 1] + 1,
                dp[i - 1, j - 1] + cost,
            )

    return dp[n, m] / max(len(a), len(b), 1)


# ------------------------------------------------------------
# Define ASR ensemble leakage score
# ------------------------------------------------------------

def compute_asr_ensemble_leakage(
    whisper_hyp,
    wav2vec_hyp,
    max_expected_words=3,
):
    """
    Real-time privacy leakage score.

    Higher = more speech leakage.

    Components:
    1. content_score:
        how much text the ASRs decoded

    2. agreement:
        whether both ASRs decoded similar content

    leakage = content_score * agreement

    Cases:
    "" and ""          -> content=0, agreement=1, leakage=0
    "RIGHT" and ""     -> content>0, agreement low, leakage low/moderate
    "RIGHT" and "RIGHT"-> content>0, agreement high, leakage high
    """

    w_words = count_words(whisper_hyp)
    v_words = count_words(wav2vec_hyp)

    avg_words = (w_words + v_words) / 2.0
    content_score = min(avg_words / max_expected_words, 1.0)

    disagreement = normalized_edit_distance_maxlen(
        whisper_hyp,
        wav2vec_hyp
    )

    agreement = 1.0 - disagreement

    leakage_score = content_score * agreement
    privacy_score = 1.0 - leakage_score

    return pd.Series({
        "whisper_words": w_words,
        "wav2vec_words": v_words,
        "avg_decoded_words": avg_words,
        "content_score": content_score,
        "asr_disagreement_fixed": disagreement,
        "asr_agreement": agreement,
        "asr_ensemble_leakage": leakage_score,
        "asr_ensemble_privacy": privacy_score,
    })


leakage_features = asr_results.apply(
    lambda r: compute_asr_ensemble_leakage(
        r["whisper_hyp"],
        r["wav2vec_hyp"],
        max_expected_words=3,   # UA-Speech mostly words/short commands
    ),
    axis=1
)

asr_results = pd.concat(
    [asr_results, leakage_features],
    axis=1
)

# clip offline metrics for plotting
for col in [
    "whisper_cer", "whisper_per",
    "wav2vec_cer", "wav2vec_per",
]:
    asr_results[col + "_clip"] = asr_results[col].clip(0, 1)


asr_results.to_csv(
    "outputs_asr_privacy/asr_results_with_realtime_leakage_score.csv",
    index=False
)
```

```python
# ============================================================
# CORRELATION: REAL-TIME SCORE VS CER/PER
# ============================================================

def compute_corr(df, x_col, y_col):
    temp = df[[x_col, y_col]].replace([np.inf, -np.inf], np.nan).dropna()

    if len(temp) < 3:
        return None

    pr, pp = pearsonr(temp[x_col], temp[y_col])
    sr, sp = spearmanr(temp[x_col], temp[y_col])

    return {
        "x": x_col,
        "y": y_col,
        "n": len(temp),
        "pearson_r": pr,
        "pearson_p": pp,
        "spearman_r": sr,
        "spearman_p": sp,
    }


corr_rows = []

for metric in [
    "whisper_cer_clip",
    "whisper_per_clip",
    "wav2vec_cer_clip",
    "wav2vec_per_clip",
]:
    corr_rows.append(
        compute_corr(
            asr_results,
            "asr_ensemble_leakage",
            metric
        )
    )

corr_df = pd.DataFrame([r for r in corr_rows if r is not None])

display(corr_df)

corr_df.to_csv(
    "outputs_asr_privacy/realtime_leakage_score_correlations.csv",
    index=False
)
```

```python
# ============================================================
# ADDITIONAL ASR PRIVACY ANALYSIS — CLEAN VERSION
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr
from sklearn.linear_model import LinearRegression

# ============================================================
# Paths
# ============================================================

ASR_RESULTS_PATH = "outputs_asr_privacy/asr_privacy_full_results.csv"

OUT_DIR = "outputs_asr_privacy/additional_analysis_figures"
os.makedirs(OUT_DIR, exist_ok=True)

# ============================================================
# Load ASR results
# ============================================================

asr_results = pd.read_csv(ASR_RESULTS_PATH)

print("Loaded ASR results:", asr_results.shape)
print("Columns:")
print(asr_results.columns.tolist())


# ============================================================
# Utility functions
# ============================================================

def safe_text(x):
    if pd.isna(x):
        return ""
    return str(x).strip()


def count_words(x):
    x = safe_text(x)
    if x == "":
        return 0
    return len(x.split())


def normalized_edit_distance_maxlen(a, b):
    """
    Correct ASR disagreement:
    - both empty => 0
    - one empty, one non-empty => 1
    - both same => 0
    """

    a = safe_text(a).replace(" ", "")
    b = safe_text(b).replace(" ", "")

    if len(a) == 0 and len(b) == 0:
        return 0.0

    n, m = len(a), len(b)

    dp = np.zeros((n + 1, m + 1), dtype=int)

    for i in range(n + 1):
        dp[i, 0] = i

    for j in range(m + 1):
        dp[0, j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):

            cost = 0 if a[i - 1] == b[j - 1] else 1

            dp[i, j] = min(
                dp[i - 1, j] + 1,
                dp[i, j - 1] + 1,
                dp[i - 1, j - 1] + cost,
            )

    return dp[n, m] / max(len(a), len(b), 1)


def clean_representation(rep):
    rep = str(rep)
    if rep == "raw":
        return "Raw"
    return rep.replace("antivad_", "")


def rep_sort_value(rep):
    rep = str(rep)
    if rep == "raw":
        return 0.0
    if "antivad_" in rep:
        return float(rep.replace("antivad_", ""))
    return 999.0


def savefig(path):
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Saved:", path)


# ============================================================
# Create missing real-time leakage columns
# ============================================================

def add_realtime_privacy_metrics(df, max_expected_words=3):
    """
    Adds:
    - whisper_words
    - wav2vec_words
    - joint_decoded_words
    - content_score
    - asr_disagreement_fixed
    - asr_agreement
    - asr_ensemble_leakage
    - asr_ensemble_privacy
    - both_empty
    """

    df = df.copy()

    required_text_cols = ["whisper_hyp", "wav2vec_hyp"]

    for col in required_text_cols:
        if col not in df.columns:
            raise ValueError(f"Missing required column: {col}")

    df["whisper_words"] = df["whisper_hyp"].apply(count_words)
    df["wav2vec_words"] = df["wav2vec_hyp"].apply(count_words)

    df["joint_decoded_words"] = (
        df["whisper_words"] + df["wav2vec_words"]
    ) / 2.0

    df["content_score"] = (
        df["joint_decoded_words"] / max_expected_words
    ).clip(0, 1)

    df["asr_disagreement_fixed"] = df.apply(
        lambda r: normalized_edit_distance_maxlen(
            r["whisper_hyp"],
            r["wav2vec_hyp"]
        ),
        axis=1
    )

    df["asr_agreement"] = 1.0 - df["asr_disagreement_fixed"]

    # Main proposed online leakage score
    df["asr_ensemble_leakage"] = (
        df["content_score"] * df["asr_agreement"]
    )

    df["asr_ensemble_privacy"] = 1.0 - df["asr_ensemble_leakage"]

    df["both_empty"] = (
        (df["whisper_words"] == 0) &
        (df["wav2vec_words"] == 0)
    ).astype(int)

    df["either_empty"] = (
        (df["whisper_words"] == 0) |
        (df["wav2vec_words"] == 0)
    ).astype(int)

    return df


asr_results = add_realtime_privacy_metrics(
    asr_results,
    max_expected_words=3
)

print("\nAfter adding real-time privacy metrics:")
print(asr_results[
    [
        "representation",
        "whisper_hyp",
        "wav2vec_hyp",
        "whisper_words",
        "wav2vec_words",
        "asr_disagreement_fixed",
        "asr_ensemble_leakage",
        "asr_ensemble_privacy",
        "both_empty",
    ]
].head())


# ============================================================
# Clip CER/PER/WER for plotting
# ============================================================

for col in [
    "whisper_cer",
    "whisper_per",
    "whisper_wer",
    "wav2vec_cer",
    "wav2vec_per",
    "wav2vec_wer",
]:
    if col in asr_results.columns:
        asr_results[col + "_clip"] = asr_results[col].clip(0, 1)

# save fixed full file
fixed_path = "outputs_asr_privacy/asr_privacy_full_results_with_online_metrics.csv"
asr_results.to_csv(fixed_path, index=False)
print("Saved:", fixed_path)


# ============================================================
# Build summaries AFTER adding columns
# ============================================================

summary_cols = [
    "whisper_cer_clip",
    "whisper_per_clip",
    "whisper_wer_clip",
    "wav2vec_cer_clip",
    "wav2vec_per_clip",
    "wav2vec_wer_clip",
    "asr_ensemble_leakage",
    "asr_ensemble_privacy",
    "asr_disagreement_fixed",
    "asr_agreement",
    "joint_decoded_words",
    "whisper_words",
    "wav2vec_words",
    "both_empty",
    "either_empty",
    "residual_energy_ratio",
    "zero_fraction",
]

summary_cols = [c for c in summary_cols if c in asr_results.columns]

asr_summary = (
    asr_results
    .groupby("representation")[summary_cols]
    .mean()
    .reset_index()
)

severity_cols = summary_cols.copy()

asr_severity = (
    asr_results
    .groupby(["severity", "representation"])[severity_cols]
    .mean()
    .reset_index()
)

summary_path = "outputs_asr_privacy/asr_privacy_summary_with_online_metrics.csv"
severity_path = "outputs_asr_privacy/asr_privacy_summary_by_severity_with_online_metrics.csv"

asr_summary.to_csv(summary_path, index=False)
asr_severity.to_csv(severity_path, index=False)

print("Saved:", summary_path)
print("Saved:", severity_path)

display(asr_summary)
display(asr_severity)


# ============================================================
# Plot style
# ============================================================

plt.rcParams.update({
    "font.size": 17,
    "axes.titlesize": 20,
    "axes.labelsize": 18,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "legend.fontsize": 13,
    "axes.linewidth": 1.8,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


# ============================================================
# 1. Threshold trend plots
# ============================================================

def prepare_threshold_df(df):
    out = df.copy()
    out["rep_label"] = out["representation"].apply(clean_representation)
    out["rep_order"] = out["representation"].apply(rep_sort_value)
    out = out.sort_values("rep_order")
    return out


def plot_threshold_metrics(df, metrics, labels, ylabel, title, save_name, ylim=None):
    df = prepare_threshold_df(df)

    plt.figure(figsize=(9, 5))

    for metric, label in zip(metrics, labels):
        if metric not in df.columns:
            print("Skipping missing metric:", metric)
            continue

        plt.plot(
            df["rep_label"],
            df[metric],
            marker="o",
            linewidth=2.8,
            label=label
        )

        for _, r in df.iterrows():
            plt.text(
                r["rep_label"],
                r[metric] + 0.02,
                f"{r[metric]:.2f}",
                ha="center",
                fontsize=10,
                fontweight="bold"
            )

    plt.xlabel("Representation / Anti-VAD Threshold")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.legend(frameon=False, ncol=2)

    if ylim:
        plt.ylim(*ylim)

    savefig(os.path.join(OUT_DIR, save_name))


plot_threshold_metrics(
    asr_summary,
    metrics=[
        "whisper_cer_clip",
        "wav2vec_cer_clip",
        "whisper_per_clip",
        "wav2vec_per_clip",
    ],
    labels=[
        "Whisper CER",
        "Wav2Vec CER",
        "Whisper PER",
        "Wav2Vec PER",
    ],
    ylabel="Error Rate",
    title="ASR Error Increases Under Anti-VAD Filtering",
    save_name="fig_threshold_cer_per.pdf",
    ylim=(0, 1.08)
)


plot_threshold_metrics(
    asr_summary,
    metrics=[
        "asr_ensemble_leakage",
        "asr_ensemble_privacy",
        "asr_disagreement_fixed",
    ],
    labels=[
        "Online Leakage Score",
        "Online Privacy Score",
        "ASR Disagreement",
    ],
    ylabel="Score",
    title="Online Privacy Metrics Across Anti-VAD Thresholds",
    save_name="fig_threshold_online_metrics.pdf",
    ylim=(0, 1.08)
)


plot_threshold_metrics(
    asr_summary,
    metrics=[
        "joint_decoded_words",
        "whisper_words",
        "wav2vec_words",
    ],
    labels=[
        "Avg. Decoded Words",
        "Whisper Words",
        "Wav2Vec Words",
    ],
    ylabel="Decoded Words",
    title="Decoded Linguistic Content Across Anti-VAD Thresholds",
    save_name="fig_threshold_decoded_words.pdf",
)


# ============================================================
# 2. Severity-wise plots
# ============================================================

def plot_severity_grouped_bars(df, metric, ylabel, title, save_name):
    plot_df = df.copy()

    reps_to_show = ["raw", "antivad_0.7", "antivad_0.8", "antivad_0.9"]
    plot_df = plot_df[plot_df["representation"].isin(reps_to_show)]

    severity_order = ["control", "high", "medium", "low", "very_low"]
    rep_map = {
        "raw": "Raw",
        "antivad_0.7": "0.7",
        "antivad_0.8": "0.8",
        "antivad_0.9": "0.9",
    }

    plot_df["severity"] = pd.Categorical(
        plot_df["severity"],
        categories=severity_order,
        ordered=True
    )

    plot_df["rep_label"] = plot_df["representation"].map(rep_map)

    pivot = plot_df.pivot(
        index="severity",
        columns="rep_label",
        values=metric
    )

    pivot = pivot[["Raw", "0.7", "0.8", "0.9"]]

    ax = pivot.plot(
        kind="bar",
        figsize=(10, 5),
        width=0.72,
        edgecolor="black",
        linewidth=1.4
    )

    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.2f",
            fontsize=9,
            fontweight="bold",
            padding=2
        )

    plt.ylabel(ylabel)
    plt.xlabel("Severity")
    plt.title(title)
    plt.xticks(rotation=20, ha="right")
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.legend(title="Representation", frameon=False, ncol=4)

    savefig(os.path.join(OUT_DIR, save_name))


plot_severity_grouped_bars(
    asr_severity,
    metric="asr_ensemble_leakage",
    ylabel="Online Leakage Score",
    title="Severity-wise Real-time Speech Leakage",
    save_name="fig_severity_online_leakage.pdf"
)

plot_severity_grouped_bars(
    asr_severity,
    metric="joint_decoded_words",
    ylabel="Mean Decoded Words",
    title="Severity-wise Decoded Linguistic Content",
    save_name="fig_severity_decoded_words.pdf"
)

plot_severity_grouped_bars(
    asr_severity,
    metric="whisper_cer_clip",
    ylabel="Whisper CER",
    title="Severity-wise Whisper CER",
    save_name="fig_severity_whisper_cer.pdf"
)

plot_severity_grouped_bars(
    asr_severity,
    metric="wav2vec_cer_clip",
    ylabel="Wav2Vec CER",
    title="Severity-wise Wav2Vec CER",
    save_name="fig_severity_wav2vec_cer.pdf"
)


# ============================================================
# 3. Correlation: online metric vs CER/PER
# ============================================================

def compute_corr(df, x_col, y_col):
    temp = df[[x_col, y_col]].replace([np.inf, -np.inf], np.nan).dropna()

    if len(temp) < 3:
        return None

    pr, pp = pearsonr(temp[x_col], temp[y_col])
    sr, sp = spearmanr(temp[x_col], temp[y_col])

    return {
        "x": x_col,
        "y": y_col,
        "n": len(temp),
        "pearson_r": pr,
        "pearson_p": pp,
        "spearman_r": sr,
        "spearman_p": sp,
    }


corr_rows = []

for target in [
    "whisper_cer_clip",
    "whisper_per_clip",
    "wav2vec_cer_clip",
    "wav2vec_per_clip",
]:
    corr_rows.append(
        compute_corr(
            asr_results,
            "asr_ensemble_leakage",
            target
        )
    )

corr_df = pd.DataFrame([r for r in corr_rows if r is not None])
corr_path = "outputs_asr_privacy/online_leakage_correlations.csv"
corr_df.to_csv(corr_path, index=False)

print("Saved:", corr_path)
display(corr_df)


# ============================================================
# 4. Scatter plots with regression line
# ============================================================

def plot_scatter_metric(df, x_col, y_col, x_label, y_label, title, save_name):
    plot_df = df[[x_col, y_col, "representation"]].replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    plt.figure(figsize=(7, 5))

    reps = sorted(plot_df["representation"].unique(), key=rep_sort_value)

    for rep in reps:
        sub = plot_df[plot_df["representation"] == rep]

        plt.scatter(
            sub[x_col],
            sub[y_col],
            s=55,
            edgecolor="black",
            linewidth=0.8,
            alpha=0.75,
            label=clean_representation(rep)
        )

    X = plot_df[[x_col]].values
    y = plot_df[y_col].values

    model = LinearRegression()
    model.fit(X, y)

    xs = np.linspace(plot_df[x_col].min(), plot_df[x_col].max(), 100)
    ys = model.predict(xs.reshape(-1, 1))

    plt.plot(xs, ys, linewidth=2.5)

    pr, pp = pearsonr(plot_df[x_col], plot_df[y_col])
    sr, sp = spearmanr(plot_df[x_col], plot_df[y_col])

    plt.xlabel(x_label)
    plt.ylabel(y_label)
    plt.title(f"{title}\nPearson r={pr:.2f}, Spearman r={sr:.2f}")

    plt.grid(True, linestyle="--", alpha=0.35)
    plt.legend(frameon=False, fontsize=10, ncol=2)

    if "clip" in y_col:
        plt.xlim(-0.03, 1.03)
        plt.ylim(-0.03, 1.03)

    savefig(os.path.join(OUT_DIR, save_name))


plot_scatter_metric(
    asr_results,
    x_col="asr_ensemble_leakage",
    y_col="whisper_cer_clip",
    x_label="Online ASR Ensemble Leakage Score",
    y_label="Whisper CER",
    title="Online Leakage Score Predicts Whisper CER",
    save_name="fig_online_leakage_vs_whisper_cer.pdf"
)

plot_scatter_metric(
    asr_results,
    x_col="asr_ensemble_leakage",
    y_col="whisper_per_clip",
    x_label="Online ASR Ensemble Leakage Score",
    y_label="Whisper PER",
    title="Online Leakage Score Predicts Whisper PER",
    save_name="fig_online_leakage_vs_whisper_per.pdf"
)

plot_scatter_metric(
    asr_results,
    x_col="asr_ensemble_leakage",
    y_col="wav2vec_cer_clip",
    x_label="Online ASR Ensemble Leakage Score",
    y_label="Wav2Vec CER",
    title="Online Leakage Score Predicts Wav2Vec CER",
    save_name="fig_online_leakage_vs_wav2vec_cer.pdf"
)

plot_scatter_metric(
    asr_results,
    x_col="asr_ensemble_leakage",
    y_col="wav2vec_per_clip",
    x_label="Online ASR Ensemble Leakage Score",
    y_label="Wav2Vec PER",
    title="Online Leakage Score Predicts Wav2Vec PER",
    save_name="fig_online_leakage_vs_wav2vec_per.pdf"
)


# ============================================================
# 5. Compare disagreement-only vs ensemble metric
# ============================================================

def compare_proxy_metrics(df):
    proxy_cols = [
        ("asr_disagreement_fixed", "Disagreement Only"),
        ("joint_decoded_words", "Decoded Words Only"),
        ("asr_ensemble_leakage", "Ensemble Leakage"),
        ("asr_ensemble_privacy", "Ensemble Privacy"),
    ]

    targets = [
        ("whisper_cer_clip", "Whisper CER"),
        ("whisper_per_clip", "Whisper PER"),
        ("wav2vec_cer_clip", "Wav2Vec CER"),
        ("wav2vec_per_clip", "Wav2Vec PER"),
    ]

    rows = []

    for proxy_col, proxy_name in proxy_cols:
        for target_col, target_name in targets:

            if proxy_col not in df.columns or target_col not in df.columns:
                continue

            temp = df[[proxy_col, target_col]].replace(
                [np.inf, -np.inf],
                np.nan
            ).dropna()

            if len(temp) < 3:
                continue

            pr, pp = pearsonr(temp[proxy_col], temp[target_col])
            sr, sp = spearmanr(temp[proxy_col], temp[target_col])

            rows.append({
                "proxy": proxy_name,
                "target": target_name,
                "pearson_r": pr,
                "pearson_abs": abs(pr),
                "pearson_p": pp,
                "spearman_r": sr,
                "spearman_abs": abs(sr),
                "spearman_p": sp,
                "n": len(temp),
            })

    out = pd.DataFrame(rows)

    out_path = "outputs_asr_privacy/proxy_metric_correlation_comparison.csv"
    out.to_csv(out_path, index=False)

    print("Saved:", out_path)
    display(out)

    return out


proxy_corr_df = compare_proxy_metrics(asr_results)


def plot_proxy_comparison(proxy_corr_df, target="Whisper CER"):
    df = proxy_corr_df[proxy_corr_df["target"] == target].copy()
    df = df.sort_values("pearson_abs", ascending=False)

    fig, ax = plt.subplots(figsize=(8, 4.8))

    bars = ax.bar(
        df["proxy"],
        df["pearson_abs"],
        width=0.55,
        edgecolor="black",
        linewidth=1.5
    )

    for bar in bars:
        h = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            h + 0.015,
            f"{h:.2f}",
            ha="center",
            fontsize=12,
            fontweight="bold"
        )

    plt.ylabel("|Pearson Correlation|")
    plt.xlabel("Real-time Proxy")
    plt.title(f"Which Online Proxy Best Predicts {target}?")
    plt.xticks(rotation=20, ha="right")
    plt.ylim(0, 1.05)
    plt.grid(axis="y", linestyle="--", alpha=0.4)

    savefig(
        os.path.join(
            OUT_DIR,
            f"fig_proxy_comparison_{target.replace(' ', '_')}.pdf"
        )
    )


plot_proxy_comparison(proxy_corr_df, "Whisper CER")
plot_proxy_comparison(proxy_corr_df, "Whisper PER")
plot_proxy_comparison(proxy_corr_df, "Wav2Vec CER")
plot_proxy_comparison(proxy_corr_df, "Wav2Vec PER")


# ============================================================
# 6. Empty transcript analysis
# ============================================================

empty_summary = (
    asr_results
    .groupby("representation")
    [["both_empty", "either_empty"]]
    .mean()
    .reset_index()
)

empty_summary_path = "outputs_asr_privacy/empty_transcript_summary.csv"
empty_summary.to_csv(empty_summary_path, index=False)

print("Saved:", empty_summary_path)
display(empty_summary)


plot_threshold_metrics(
    empty_summary,
    metrics=["both_empty", "either_empty"],
    labels=["Both ASRs Empty", "Either ASR Empty"],
    ylabel="Fraction of Files",
    title="Empty ASR Outputs Increase Under Privacy Filtering",
    save_name="fig_empty_asr_outputs.pdf",
    ylim=(0, 1.05)
)


# ============================================================
# 7. Final online/offline privacy table
# ============================================================

final_cols = [
    "representation",
    "asr_ensemble_leakage",
    "asr_ensemble_privacy",
    "joint_decoded_words",
    "whisper_cer_clip",
    "whisper_per_clip",
    "wav2vec_cer_clip",
    "wav2vec_per_clip",
    "both_empty",
    "residual_energy_ratio",
    "zero_fraction",
]

final_cols = [c for c in final_cols if c in asr_summary.columns]

online_offline_summary = asr_summary[final_cols].copy()
online_offline_summary["rep_label"] = online_offline_summary["representation"].apply(clean_representation)
online_offline_summary["rep_order"] = online_offline_summary["representation"].apply(rep_sort_value)
online_offline_summary = online_offline_summary.sort_values("rep_order")

out_path = "outputs_asr_privacy/online_offline_privacy_summary_final.csv"
online_offline_summary.to_csv(out_path, index=False)

print("Saved:", out_path)
display(online_offline_summary)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

OUT_DIR = "outputs_asr_privacy/speech_leakage_figures_clean"
os.makedirs(OUT_DIR, exist_ok=True)

plt.rcParams.update({
    "font.size": 16,
    "axes.titlesize": 20,
    "axes.labelsize": 18,
    "xtick.labelsize": 15,
    "ytick.labelsize": 15,
    "legend.fontsize": 14,
    "axes.linewidth": 1.8,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})


def savefig(path):
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    print("Saved:", path)


def clean_rep(rep):
    if rep == "raw":
        return "Raw"
    return rep.replace("antivad_", "")


def rep_order(rep):
    if rep == "raw":
        return 0
    return float(rep.replace("antivad_", ""))


def prepare_rep_order(df):
    df = df.copy()
    df["rep_label"] = df["representation"].apply(clean_rep)
    df["rep_order"] = df["representation"].apply(rep_order)
    return df.sort_values("rep_order")
```

```python
def plot_threshold_cer_only(threshold_df):
    df = prepare_rep_order(threshold_df)

    df["whisper_cer_clip"] = df["whisper_cer"].clip(0, 1)
    df["wav2vec_cer_clip"] = df["wav2vec_cer"].clip(0, 1)

    fig, ax = plt.subplots(figsize=(8.2, 4.6))

    ax.plot(
        df["rep_label"],
        df["whisper_cer_clip"],
        marker="o",
        linewidth=3,
        markersize=8,
        label="Whisper CER"
    )

    ax.plot(
        df["rep_label"],
        df["wav2vec_cer_clip"],
        marker="s",
        linewidth=3,
        markersize=8,
        label="Wav2Vec CER"
    )

    # annotate every other point lightly to avoid collisions
    for i, r in df.iterrows():
        ax.text(
            r["rep_label"],
            min(r["whisper_cer_clip"] + 0.035, 1.04),
            f"{r['whisper_cer_clip']:.2f}",
            ha="center",
            fontsize=10,
            fontweight="bold"
        )

        ax.text(
            r["rep_label"],
            max(r["wav2vec_cer_clip"] - 0.055, 0.02),
            f"{r['wav2vec_cer_clip']:.2f}",
            ha="center",
            fontsize=10,
            fontweight="bold"
        )

    ax.set_xlabel("Representation / Anti-VAD Threshold")
    ax.set_ylabel("CER")
    ax.set_title("ASR Speech Leakage Across Anti-VAD Thresholds")

    ax.set_ylim(0, 1.08)
    ax.grid(axis="y", linestyle="--", alpha=0.35)

    ax.legend(
        frameon=False,
        loc="lower right"
    )

    savefig(os.path.join(OUT_DIR, "fig_threshold_cer_only_clean.pdf"))


plot_threshold_cer_only(threshold_df)
```

```python
def plot_severity_wav2vec_cer_clean(severity_df):
    df = severity_df.copy()

    reps_to_show = ["raw", "antivad_0.7", "antivad_0.8", "antivad_0.9"]
    df = df[df["representation"].isin(reps_to_show)].copy()

    severity_order = ["control", "high", "medium", "low", "very_low"]
    rep_map = {
        "raw": "Raw",
        "antivad_0.7": "0.7",
        "antivad_0.8": "0.8",
        "antivad_0.9": "0.9",
    }

    df["wav2vec_cer_clip"] = df["wav2vec_cer"].clip(0, 1)
    df["severity"] = pd.Categorical(df["severity"], categories=severity_order, ordered=True)
    df["rep_label"] = df["representation"].map(rep_map)

    pivot = df.pivot(
        index="severity",
        columns="rep_label",
        values="wav2vec_cer_clip"
    )

    pivot = pivot[["Raw", "0.7", "0.8", "0.9"]]

    fig, ax = plt.subplots(figsize=(9.2, 4.8))

    pivot.plot(
        kind="bar",
        ax=ax,
        width=0.68,
        edgecolor="black",
        linewidth=1.4
    )

    # annotate bars, smaller and rotated slightly
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.2f",
            fontsize=8,
            fontweight="bold",
            padding=2,
            rotation=0
        )

    ax.set_xlabel("Intelligibility")
    ax.set_ylabel("Wav2Vec CER")
    # ax.set_title("Severity-wise Speech Leakage")

    ax.set_ylim(0, 1.10)
    ax.set_xticklabels(
        ["Control", "High", "Medium", "Low", "Very Low"],
        rotation=0,
        ha="right"
    )

    ax.grid(axis="y", linestyle="--", alpha=0.35)

    # legend above plot, not inside
    ax.legend(
        # title="Representation",
        frameon=False,
        ncol=4,
        loc="upper center",
        bbox_to_anchor=(0.5, 1.18)
    )

    savefig(os.path.join(OUT_DIR, "fig_severity_wav2vec_cer_clean.pdf"))


plot_severity_wav2vec_cer_clean(severity_df)
```

```python
# ============================================================
# Speaker Identity + Gender Leakage Evaluation
# Raw vs AntiVAD-0.7
# Feature-group leakage ranking
# ============================================================

import re
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold, StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
from sklearn.inspection import permutation_importance


# -----------------------------
# Config
# -----------------------------
CSV_PATH = "features_rich (3).csv"

REPRESENTATIONS = {
    "Raw": "raw",
    "AntiVAD-0.7": "silero_antivad_0.7",
    "AntiVAD-0.8": "silero_antivad_0.8",
    "AntiVAD-0.9": "silero_antivad_0.9"
}

META_COLS = [
    "audio_path", "speaker_id", "label", "label_3class",
    "representation", "duration_sec", "residual_energy_ratio",
    "zero_fraction"
]

RANDOM_STATE = 42
N_SPLITS = 5


# -----------------------------
# Load data
# -----------------------------
df = pd.read_csv(CSV_PATH)

# infer gender from UA-Speech speaker IDs:
# CF02 = control female, CM04 = control male,
# F02 = dysarthric female, M04 = dysarthric male
def infer_gender(speaker_id):
    sid = str(speaker_id).upper()
    if re.match(r"^C?F", sid):
        return "female"
    elif re.match(r"^C?M", sid):
        return "male"
    else:
        return np.nan

if "gender" not in df.columns:
    df["gender"] = df["speaker_id"].apply(infer_gender)

feature_cols = [
    c for c in df.columns
    if c not in META_COLS + ["gender"]
    and pd.api.types.is_numeric_dtype(df[c])
]

print(f"Total rows: {len(df)}")
print(f"Total speakers: {df['speaker_id'].nunique()}")
print(f"Total features: {len(feature_cols)}")
print(df["representation"].value_counts())


# -----------------------------
# Feature group mapping
# -----------------------------
def get_feature_group(col):
    if "__" in col:
        return col.split("__")[0]
    else:
        return "other"

feature_groups = {}
for c in feature_cols:
    g = get_feature_group(c)
    feature_groups.setdefault(g, []).append(c)

print("\nFeature groups:")
for g, cols in feature_groups.items():
    print(f"{g}: {len(cols)} features")


# -----------------------------
# Model
# -----------------------------
def make_model():
    return Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("clf", RandomForestClassifier(
            n_estimators=500,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ])


# -----------------------------
# Generic CV evaluator
# -----------------------------
def evaluate_task(data, target_col, features, group_col=None, task_name=""):
    X = data[features].replace([np.inf, -np.inf], np.nan)
    y_raw = data[target_col].astype(str)

    le = LabelEncoder()
    y = le.fit_transform(y_raw)

    if len(np.unique(y)) < 2:
        return None

    if group_col is not None:
        groups = data[group_col].astype(str).values
        splitter = StratifiedGroupKFold(
            n_splits=N_SPLITS,
            shuffle=True,
            random_state=RANDOM_STATE
        )
        split_iter = splitter.split(X, y, groups)
    else:
        splitter = StratifiedKFold(
            n_splits=N_SPLITS,
            shuffle=True,
            random_state=RANDOM_STATE
        )
        split_iter = splitter.split(X, y)

    accs, baccs, f1s = [], [], []

    for train_idx, test_idx in split_iter:
        model = make_model()
        model.fit(X.iloc[train_idx], y[train_idx])
        pred = model.predict(X.iloc[test_idx])

        accs.append(accuracy_score(y[test_idx], pred))
        baccs.append(balanced_accuracy_score(y[test_idx], pred))
        f1s.append(f1_score(y[test_idx], pred, average="macro"))

    return {
        "task": task_name,
        "target": target_col,
        "n_classes": len(le.classes_),
        "accuracy_mean": np.mean(accs),
        "accuracy_std": np.std(accs),
        "balanced_accuracy_mean": np.mean(baccs),
        "balanced_accuracy_std": np.std(baccs),
        "macro_f1_mean": np.mean(f1s),
        "macro_f1_std": np.std(f1s),
    }


# -----------------------------
# Main leakage evaluation
# -----------------------------
results = []

for rep_name, rep_value in REPRESENTATIONS.items():
    drep = df[df["representation"] == rep_value].copy()

    # Speaker ID leakage:
    # random stratified folds over utterances.
    speaker_res = evaluate_task(
        drep,
        target_col="speaker_id",
        features=feature_cols,
        group_col=None,
        task_name=f"{rep_name} speaker identity leakage"
    )
    speaker_res["representation"] = rep_name
    results.append(speaker_res)

    # Gender leakage:
    # speaker-independent split to avoid memorizing speaker identity.
    gender_res = evaluate_task(
        drep.dropna(subset=["gender"]),
        target_col="gender",
        features=feature_cols,
        group_col="speaker_id",
        task_name=f"{rep_name} gender leakage"
    )
    gender_res["representation"] = rep_name
    results.append(gender_res)

results_df = pd.DataFrame(results)
print("\n=== Overall leakage results ===")
print(results_df)

results_df.to_csv("privacy_leakage_raw_vs_antivad07.csv", index=False)


# -----------------------------
# Feature-group leakage ranking
# -----------------------------
group_results = []

for rep_name, rep_value in REPRESENTATIONS.items():
    drep = df[df["representation"] == rep_value].copy()

    for group_name, group_cols in feature_groups.items():

        # Speaker leakage from one group
        speaker_group_res = evaluate_task(
            drep,
            target_col="speaker_id",
            features=group_cols,
            group_col=None,
            task_name="speaker_identity"
        )
        if speaker_group_res is not None:
            speaker_group_res["representation"] = rep_name
            speaker_group_res["feature_group"] = group_name
            speaker_group_res["n_features"] = len(group_cols)
            group_results.append(speaker_group_res)

        # Gender leakage from one group
        gender_group_res = evaluate_task(
            drep.dropna(subset=["gender"]),
            target_col="gender",
            features=group_cols,
            group_col="speaker_id",
            task_name="gender"
        )
        if gender_group_res is not None:
            gender_group_res["representation"] = rep_name
            gender_group_res["feature_group"] = group_name
            gender_group_res["n_features"] = len(group_cols)
            group_results.append(gender_group_res)

group_results_df = pd.DataFrame(group_results)

print("\n=== Feature-group leakage ranking ===")
print(
    group_results_df.sort_values(
        ["target", "representation", "balanced_accuracy_mean"],
        ascending=[True, True, False]
    )
)

group_results_df.to_csv("feature_group_leakage_ranking.csv", index=False)


# -----------------------------
# Permutation importance by feature group
# -----------------------------
perm_rows = []

for rep_name, rep_value in REPRESENTATIONS.items():
    drep = df[df["representation"] == rep_value].copy()

    for target_col, group_col in [
        ("speaker_id", None),
        ("gender", "speaker_id")
    ]:
        data = drep.dropna(subset=[target_col]).copy()
        X = data[feature_cols].replace([np.inf, -np.inf], np.nan)
        y_raw = data[target_col].astype(str)

        le = LabelEncoder()
        y = le.fit_transform(y_raw)

        model = make_model()
        model.fit(X, y)

        perm = permutation_importance(
            model,
            X,
            y,
            n_repeats=10,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            scoring="balanced_accuracy"
        )

        feat_imp = pd.DataFrame({
            "feature": feature_cols,
            "importance": perm.importances_mean
        })
        feat_imp["feature_group"] = feat_imp["feature"].apply(get_feature_group)

        group_imp = (
            feat_imp.groupby("feature_group")["importance"]
            .sum()
            .reset_index()
            .sort_values("importance", ascending=False)
        )

        for _, row in group_imp.iterrows():
            perm_rows.append({
                "representation": rep_name,
                "target": target_col,
                "feature_group": row["feature_group"],
                "permutation_importance_sum": row["importance"]
            })

perm_df = pd.DataFrame(perm_rows)

print("\n=== Permutation importance by feature group ===")
print(
    perm_df.sort_values(
        ["target", "representation", "permutation_importance_sum"],
        ascending=[True, True, False]
    )
)

perm_df.to_csv("feature_group_permutation_importance.csv", index=False)


# -----------------------------
# Optional: quick paper-ready summary
# -----------------------------
print("\nSaved files:")
print("1. privacy_leakage_raw_vs_antivad07.csv")
print("2. feature_group_leakage_ranking.csv")
print("3. feature_group_permutation_importance.csv")
```
